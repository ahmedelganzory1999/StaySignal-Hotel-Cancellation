# Data Cleaning

In this part, I will go through the problems found in the raw data. I will only change a value when I have a reasonable reason to do so. I will also check the data again after cleaning.

## 14. Create a Working Copy

```python
# Work on a copy and keep the raw data unchanged
df = raw_df.copy()
rows_before = len(df)

print("Rows before cleaning:", rows_before)
print("Missing values before cleaning:", df.isna().sum().sum())
print("Fully identical rows:", df.duplicated().sum())
```

## 15. Handling Missing Values

I will handle each column separately instead of using the same method for all missing values.

```python
missing_before = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing %": (df.isna().mean() * 100).round(2)
})
missing_before[missing_before["Missing Count"] > 0].sort_values("Missing Count", ascending=False)
```

### 15.1 Missing Children Values

There are only four missing values in `children`. All four bookings have a recorded number of adults, but there is no way to know the actual number of children. I will remove these four rows rather than assume that the answer is zero.

```python
# Check the four records before removing them
df.loc[df["children"].isna(), [
    "hotel", "is_canceled", "adults", "children", "babies", "country"
]]
```

```python
rows_missing_children = df["children"].isna().sum()
df = df.dropna(subset=["children"]).copy()
df["children"] = df["children"].astype(int)

print("Removed rows:", rows_missing_children)
print("Rows remaining:", len(df))
```

### 15.2 Missing Country Values

For `country`, I cannot work out the correct country from the other columns. I will use `Unknown` so those bookings stay in the dataset without assigning them a country.

```python
df["country"] = df["country"].fillna("Unknown")
print("Missing country values:", df["country"].isna().sum())
print("Unknown country records:", (df["country"] == "Unknown").sum())
```

### 15.3 Missing Agent and Company IDs

`agent` and `company` are IDs, not actual numerical measurements. I will treat them as categories. When an ID is missing, I will keep it as `Unknown`. This does **not** mean that the booking had no agent or company.

`company` is missing in most of the records, so we should be careful about using this column in the ML models later.

```python
# Keep the IDs as text categories instead of numbers
for col in ["agent", "company"]:
    df[col] = df[col].astype("Int64").astype("string").fillna("Unknown")

print(df[["agent", "company"]].dtypes)
print("Unknown agents:", (df["agent"] == "Unknown").sum())
print("Unknown companies:", (df["company"] == "Unknown").sum())
```

### 15.4 Check Missing Values Again

```python
missing_after = df.isna().sum()
print("Total missing values after cleaning:", missing_after.sum())
missing_after[missing_after > 0]
```

## 16. Checking Identical Rows

I found a large number of fully identical rows, but there is no booking ID that proves these are accidental duplicates. Two different bookings can have the same information, especially if they were made through the same group or travel agent. I will keep them for now.

To see why this matters, I will compare the cancellation rate with and without these rows. The second calculation is only a test; it will not change `df`.

```python
identical_rows = df.duplicated().sum()

all_cancellation_rate = df["is_canceled"].mean() * 100
without_identical_rate = df.drop_duplicates()["is_canceled"].mean() * 100

print("Fully identical rows:", identical_rows)
print(f"Cancellation rate (all records): {all_cancellation_rate:.2f}%")
print(f"Cancellation rate (if identical rows are removed): {without_identical_rate:.2f}%")
```

Removing the identical rows would change the cancellation rate a lot. Since I cannot confirm they are repeated bookings, I will **not** drop them.

## 17. Check Guest Counts

A hotel booking would normally have at least one guest. I will look at bookings where adults, children and babies are all zero, then decide whether to remove them.

```python
# Check bookings with no guests recorded
total_guests = df["adults"] + df["children"] + df["babies"]
zero_guest_rows = total_guests == 0

print("Bookings with zero recorded guests:", zero_guest_rows.sum())
print("By hotel:")
print(df.loc[zero_guest_rows, "hotel"].value_counts())
print("By booking outcome:")
print(df.loc[zero_guest_rows, "is_canceled"].value_counts().sort_index())
```

```python
# See how many zero-guest bookings also have zero nights
nights = df["stays_in_week_nights"] + df["stays_in_weekend_nights"]

print("Zero-guest bookings with zero nights:", ((total_guests == 0) & (nights == 0)).sum())
print("Zero-guest bookings with zero ADR:", ((total_guests == 0) & (df["adr"] == 0)).sum())

df.loc[zero_guest_rows, [
    "hotel", "is_canceled", "adults", "children", "babies",
    "stays_in_week_nights", "stays_in_weekend_nights", "adr"
]].head(10)
```

There are 180 zero-guest records. Some have nights booked and even a positive room rate. This could be a recording problem or a special type of booking, but the file does not tell us which. I will **keep these records for now**, highlight the issue, and decide whether to exclude them when preparing model inputs. I do not want to delete real bookings without proof.

## 25. Lead Time

`lead_time` is the number of days between booking and planned arrival. I want to see its distribution and whether people who book earlier cancel more often.

```python
df["lead_time"].describe(percentiles=[0.25, 0.5, 0.75, 0.9, 0.95]).round(2)
```

The median lead time is **69 days**, but some reservations were made more than **700 days** before arrival. The distribution is right-skewed, so a lot of bookings are made relatively close to arrival.

### 25.1 Lead Time and Cancellation

I will divide lead time into groups. This makes the difference in cancellation rates easier to compare.

```python
lead_bins = [-1, 7, 30, 90, 180, 365, np.inf]
lead_labels = ["0-7 days", "8-30 days", "31-90 days", "91-180 days", "181-365 days", "366+ days"]

lead_group = pd.cut(df["lead_time"], bins=lead_bins, labels=lead_labels)
lead_cancel = df.groupby(lead_group, observed=True)["is_canceled"].agg(["size", "mean"]).reset_index()
lead_cancel.columns = ["Lead Time", "Bookings", "Cancellation Rate"]
lead_cancel["Cancellation Rate %"] = (lead_cancel["Cancellation Rate"] * 100).round(2)
lead_cancel[["Lead Time", "Bookings", "Cancellation Rate %"]]
```

```python
fig = px.bar(
    lead_cancel, x="Lead Time", y="Cancellation Rate %",
    text="Cancellation Rate %", title="Cancellation Rate by Lead Time Group",
    labels={"Cancellation Rate %": "Cancellation Rate (%)"}
)
fig.update_traces(texttemplate="%{text:.1f}%", textposition="outside")
fig.show()
```

The difference is quite clear. Bookings made within **7 days** have about a **9.6%** cancellation rate, compared with about **67.7%** for bookings made more than a year ahead. Lead time looks like an important variable for our later models.

# Feature Engineering and Machine Learning

Now I will prepare the features and compare the classification models.

## 38. Feature Engineering

I will create `total_nights`, `total_guests` and `has_children`. These are based on the original booking details.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import classification_report, confusion_matrix
import joblib
import json
```

```python
def add_booking_features(data):
    data = data.copy()
    data["total_nights"] = data["stays_in_week_nights"] + data["stays_in_weekend_nights"]
    data["total_guests"] = data["adults"] + data["children"] + data["babies"]
    data["has_children"] = ((data["children"] + data["babies"]) > 0).astype(int)
    return data

model_df = add_booking_features(df)
model_df[["total_nights", "total_guests", "has_children"]].describe().round(2)
```

## 39. Target and Input Features

`is_canceled` will be the target. I will use booking information that may be available before the cancellation. I will exclude `reservation_status`, `reservation_status_date`, `assigned_room_type` and other fields that may describe later events.

```python
feature_columns = [
    "hotel", "lead_time", "arrival_date_month", "arrival_date_day_of_month",
    "total_nights", "stays_in_weekend_nights", "total_guests", "has_children",
    "meal", "market_segment", "distribution_channel", "is_repeated_guest",
    "previous_cancellations", "previous_bookings_not_canceled", "reserved_room_type",
    "deposit_type", "customer_type", "adr", "total_of_special_requests",
    "required_car_parking_spaces", "country", "agent"
]
X = model_df[feature_columns].copy()
y = model_df["is_canceled"].copy()
print("Features:", X.shape[1], "Target:", y.name)
print("Missing model inputs:", X.isna().sum().sum())
```

The parking and special-request fields are very predictive in this dataset. Their exact recording time is not confirmed, so the final model has this limitation.

## 40. Training, Validation and Test Sets

I will split by arrival date. The earlier records are used for training and validation, while the latest arrivals are kept as the test period.

```python
month_numbers = {m: i for i, m in enumerate(month_order, 1)}
arrival_dates = pd.to_datetime(pd.DataFrame({
    "year": model_df["arrival_date_year"],
    "month": model_df["arrival_date_month"].map(month_numbers),
    "day": model_df["arrival_date_day_of_month"]
}, index=model_df.index))

split_date = arrival_dates.sort_values().iloc[int(len(arrival_dates) * 0.80)]
train_rows = arrival_dates < split_date
test_rows = ~train_rows
validation_date = arrival_dates[train_rows].sort_values().iloc[int(train_rows.sum() * 0.80)]
fit_rows = arrival_dates < validation_date
valid_rows = (arrival_dates >= validation_date) & train_rows

X_fit, y_fit = X.loc[fit_rows], y.loc[fit_rows]
X_valid, y_valid = X.loc[valid_rows], y.loc[valid_rows]
X_train, y_train = X.loc[train_rows], y.loc[train_rows]
X_test, y_test = X.loc[test_rows], y.loc[test_rows]

print("Training rows for model comparison:", len(X_fit))
print("Validation rows:", len(X_valid))
print("Final training rows:", len(X_train))
print("Test rows:", len(X_test))
print("Validation starts:", validation_date.date())
print("Test starts:", split_date.date())
```

```python
split_check = pd.DataFrame({
    "Bookings": [len(X_fit), len(X_valid), len(X_test)],
    "Canceled %": [y_fit.mean() * 100, y_valid.mean() * 100, y_test.mean() * 100]
}, index=["Model Training", "Validation", "Test"])
split_check.round(2)
```

The cancellation rates are not identical across the periods. I will keep that in mind when comparing models.

## 41. Preprocessing

I have numerical and categorical features, so I will prepare them separately and combine them using `ColumnTransformer` like we did in the sessions. The transformations will be fitted on training data only.

```python
categorical_features = X.select_dtypes(include=["object", "string"]).columns.tolist()
numerical_features = [col for col in feature_columns if col not in categorical_features]

print("Numerical columns:", numerical_features)
print("Categorical columns:", categorical_features)
```

### 41.1 Numerical Pipeline

I will use `StandardScaler` for numerical columns.

```python
num_pipeline = Pipeline(steps=[
    ("Scaling", StandardScaler())
])
num_pipeline
```

### 41.2 Categorical Pipeline

I will use `OneHotEncoder`. Unknown categories can appear in future bookings, so I will allow them without an error.

```python
cat_pipeline = Pipeline(steps=[
    ("Encoding", OneHotEncoder(handle_unknown="ignore"))
])
cat_pipeline
```

### 41.3 Combine the Pipelines

```python
preprocessing = ColumnTransformer(transformers=[
    ("num", num_pipeline, numerical_features),
    ("cat", cat_pipeline, categorical_features)
])
preprocessing
```

### 41.4 Complete Model Pipeline

This is the same structure we used in the lecture: preprocessing first, then the algorithm.

```python
from sklearn import set_config
set_config(display="diagram")

logistic_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessing),
    ("model", LogisticRegression(max_iter=1000, random_state=42))
])
logistic_pipeline
```

The pipeline keeps the encoding and scaling inside model training. I do not need to transform the full dataset before splitting it.

## 42. Model Comparison

I will compare Logistic Regression, Decision Tree and XGBoost. First, I will also check the result of predicting every booking as not canceled.

```python
def scores(y_true, prediction):
    return {
        "Accuracy": accuracy_score(y_true, prediction),
        "Precision": precision_score(y_true, prediction, zero_division=0),
        "Recall": recall_score(y_true, prediction, zero_division=0),
        "F1": f1_score(y_true, prediction, zero_division=0)
    }

# Use the same preprocessing for each model.
def build_pipeline(algorithm, columns=None):
    if columns is None:
        columns = feature_columns
    cat_cols = [col for col in columns if col in categorical_features]
    num_cols = [col for col in columns if col not in cat_cols]

    preprocess = ColumnTransformer(transformers=[
        ("num", Pipeline(steps=[("Scaling", StandardScaler())]), num_cols),
        ("cat", Pipeline(steps=[("Encoding", OneHotEncoder(handle_unknown="ignore"))]), cat_cols)
    ])
    return Pipeline(steps=[("preprocessor", preprocess), ("model", algorithm)])
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Decision Tree": DecisionTreeClassifier(max_depth=8, min_samples_leaf=20, random_state=42),
    "XGBoost": XGBClassifier(n_estimators=150, max_depth=5, learning_rate=0.1,
                             eval_metric="logloss", random_state=42, n_jobs=2, tree_method="hist")
}

model_results = [{"Model": "Majority Baseline", **scores(y_valid, np.zeros(len(y_valid), dtype=int))}]
fitted_models = {}
for name, algorithm in models.items():
    model_pipeline = build_pipeline(algorithm)
    model_pipeline.fit(X_fit, y_fit)
    fitted_models[name] = model_pipeline
    model_results.append({"Model": name, **scores(y_valid, model_pipeline.predict(X_valid))})

comparison = pd.DataFrame(model_results).set_index("Model").sort_values("F1", ascending=False)
comparison.round(3)
```

```python
fig = px.bar(comparison.reset_index(), x="Model", y=["Precision", "Recall", "F1"],
             barmode="group", title="Model Comparison - Validation Period")
fig.show()
```

The models do not perform equally well. I am focusing mainly on F1-score and recall, because we want to identify cancellations without producing too many false alarms.

## 43. Feature Selection

I will check Decision Tree feature importance. The categories are encoded into several columns, so I will add their importance back to the original feature name.

```python
tree_pipeline = fitted_models["Decision Tree"]
encoded_names = tree_pipeline.named_steps["preprocessor"].get_feature_names_out()
encoded_importance = tree_pipeline.named_steps["model"].feature_importances_

importance = {}
for column in feature_columns:
    if column in numerical_features:
        mask = encoded_names == "num__" + column
    else:
        mask = np.array([name.startswith("cat__" + column + "_") for name in encoded_names])
    importance[column] = encoded_importance[mask].sum()
importance_df = pd.Series(importance, name="Importance").sort_values(ascending=False)
importance_df.head(12).round(3)
```

```python
fig = px.bar(importance_df.head(12).sort_values().reset_index(),
             x="Importance", y="index", orientation="h", title="Top Original Features (Decision Tree)")
fig.show()
```

```python
top_features = importance_df.head(12).index.tolist()
small_tree = build_pipeline(
    DecisionTreeClassifier(max_depth=8, min_samples_leaf=20, random_state=42),
    top_features
)
small_tree.fit(X_fit[top_features], y_fit)
feature_selection_check = pd.DataFrame({
    "All 22 features": scores(y_valid, tree_pipeline.predict(X_valid)),
    "Top 12 features": scores(y_valid, small_tree.predict(X_valid[top_features]))
}).T
feature_selection_check.round(3)
```

The smaller feature set is only useful if its validation scores remain good. Selecting fewer columns does not always give a better model.

## 44. Hyperparameter Tuning

`GridSearchCV` checks different model settings using cross-validation. I will tune Decision Tree as a simple example from the sessions.

```python
tree_search = GridSearchCV(
    estimator=build_pipeline(DecisionTreeClassifier(random_state=42)),
    param_grid={"model__max_depth": [6, 10], "model__min_samples_leaf": [10, 30]},
    scoring="f1", cv=3, n_jobs=2
)
tree_search.fit(X_fit, y_fit)
print("Best tree parameters:", tree_search.best_params_)
print("Cross-validation F1:", round(tree_search.best_score_, 3))
print("Validation scores:", {k: round(v, 3) for k, v in scores(y_valid, tree_search.predict(X_valid)).items()})
```

### 44.1 Stronger XGBoost Candidate

I will compare the first XGBoost setup with a stronger configuration. I will still choose the prediction threshold on the validation records, not the test period.

```python
final_params = {
    "n_estimators": 400, "max_depth": 7, "learning_rate": 0.05,
    "min_child_weight": 4, "subsample": 0.9, "colsample_bytree": 0.9
}

strong_xgb = build_pipeline(XGBClassifier(
    eval_metric="logloss", random_state=42, n_jobs=2, tree_method="hist", **final_params
))
strong_xgb.fit(X_fit, y_fit)

candidate_check = pd.DataFrame({
    "Initial XGBoost (threshold 0.5)": scores(y_valid, fitted_models["XGBoost"].predict(X_valid)),
    "Stronger XGBoost (threshold 0.5)": scores(y_valid, strong_xgb.predict(X_valid))
}).T
candidate_check.round(3)
```

### 44.2 Choosing the Prediction Threshold

By default, the model uses a 0.50 threshold. I will check a few different values on validation data and select the one with the highest F1-score.

```python
validation_prob = strong_xgb.predict_proba(X_valid)[:, 1]
threshold_results = []
for threshold_value in np.arange(0.20, 0.61, 0.05):
    prediction = (validation_prob >= threshold_value).astype(int)
    threshold_results.append({"Threshold": round(float(threshold_value), 2),
                              **scores(y_valid, prediction)})
threshold_table = pd.DataFrame(threshold_results)
threshold_table.round(3)
```

```python
best_row = threshold_table.sort_values(["F1", "Recall"], ascending=False).iloc[0]
threshold = float(best_row["Threshold"])
print("Selected threshold:", threshold)
print("Validation scores:")
print(best_row[["Accuracy", "Precision", "Recall", "F1"]].round(3))

fig = px.line(threshold_table, x="Threshold", y=["Precision", "Recall", "F1"],
              markers=True, title="Choosing the Threshold (Validation Data)")
fig.show()
```

Changing the threshold affects precision and recall. I will keep the threshold selected using validation F1-score.

## 45. Final Model Evaluation

I will now fit the selected pipeline on the whole earlier period, then test it on the later period.

```python
final_model = build_pipeline(XGBClassifier(
    eval_metric="logloss", random_state=42, n_jobs=2, tree_method="hist", **final_params
))
final_model.fit(X_train, y_train)
test_probability = final_model.predict_proba(X_test)[:, 1]
test_prediction = (test_probability >= threshold).astype(int)
final_scores = scores(y_test, test_prediction)
pd.DataFrame([final_scores], index=["Final XGBoost"]).round(3)
```

```python
print(classification_report(y_test, test_prediction,
                            target_names=["Not Canceled", "Canceled"], zero_division=0))
cm = confusion_matrix(y_test, test_prediction, labels=[0, 1])
fig = px.imshow(cm, text_auto=True, x=["Predicted No", "Predicted Yes"],
                y=["Actual No", "Actual Yes"], title="Test Confusion Matrix")
fig.show()
print("Confusion matrix:", cm.tolist())
```
